## Ajout des colonnes
Pour les 4 set, on voit que les Identifiant machines sont tous nommé de 1 à X. 
Ce qui veut dire que pour assembler 2 sets il faut ajotuer un point de distinction notament, le numéro de la ligne.

Dans le site du dataset on apprends les types de pannes et conditions de chaque trains : 

De ce postula, Il y a 2 choix : 

Option A — Regrouper par type de machine (mode de panne)
- Usine 1 : FD001 + FD002 (machines qui tombent en panne sur le compresseur uniquement)
- Usine 2 : FD003 + FD004 (machines avec 2 composants critiques)
- Argument : dans une vraie usine, les machines d'une même ligne font le même travail → même type d'usure

Option B — Regrouper par environnement de travail (conditions opérationnelles)
- Usine 1 : FD001 + FD003 (régime stable → conditions de production fixes, ex : pièces aéronautiques très normalisées)
- Usine 2 : FD002 + FD004 (multi-régimes → production variée, changements fréquents, ex : pièces automobile en série variable)
- Argument : une usine automobile vs une usine aéronautique n'a pas les mêmes régimes de fonctionnement

Etant donnée que Mecha est une entreprise avec 2 types de domaines, je pense que l'option B est plus cohérentes. 

```python
# Premier essaie, création de la colonne Usine et Ligne
def Creation_Colonne(Liste, Usine, Ligne):
    for fichier in Liste:
        fichier['usine'] = Usine
        fichier['ligne_production'] = Ligne
```

```python
def Creation_Colonne(Liste, Usine, Ligne):
    for fichier in Liste:
        fichier['usine'] = Usine
        fichier['ligne_production'] = Ligne
        fichier['machine_uid'] = fichier['machine_id'].astype(str) + '_' + str(Usine) + '_' + str(Ligne)

# FD001 (Usine 1, Ligne 1)
FD001 = [DF_Train_01, DF_Tests_01, DF_Rul_01]
Creation_Colonne(FD001, 1, 1)

# FD003 (Usine 1, Ligne 2)
FD003 = [DF_Train_03, DF_Tests_03, DF_Rul_03]
Creation_Colonne(FD003, 1, 2)

# FD002 (Usine 2, Ligne 1)
FD002 = [DF_Train_02, DF_Tests_02, DF_Rul_02]
Creation_Colonne(FD002, 2, 1)

# FD004 (Usine 2, Ligne 2)
FD004 = [DF_Train_04, DF_Tests_04, DF_Rul_04]
Creation_Colonne(FD004, 2, 2)

DF_Train_01
```

Maintenant que les colonnes sont mise en place et les sets séparé. Alors on peut les assemblé

```python
# Assemblage des trains

train_usine1 = pd.concat([DF_Train_01, DF_Train_03], ignore_index=True)
test_usine1  = pd.concat([DF_Tests_01, DF_Tests_03],  ignore_index=True)
rul_usine1   = pd.concat([DF_Rul_01, DF_Rul_03],   ignore_index=True)

train_usine2 = pd.concat([DF_Train_02, DF_Train_04], ignore_index=True)
test_usine2  = pd.concat([DF_Tests_02, DF_Tests_04],  ignore_index=True)
rul_usine2   = pd.concat([DF_Rul_02, DF_Rul_04],   ignore_index=True)

train_usine1
```
